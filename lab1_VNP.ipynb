Import all libraries and modules that you are going to use.

```
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
```

# <font color = 'Orange'> Start the scraping by sending HTTP request</font> (0.3 POINTS)


 **Fetch a website's content - send request**

```
url = "https://mobelix.com.mk/mk/mobilni-telefoni"
response = requests.get(url)
```

**Get the response-status code and see:**
<li>If the code starts with a 2, it means the request was successful.



```
if response.status_code // 100 == 2:
    print("Request was successful.")
else:
    print(f"Request failed with status code: {response.status_code}")
```

# <font color = 'orange'>Parse the HTML with BeautifulSoup</font> (0.3 POINTS)

Beautiful Soup is a Python library for pulling data out of HTML and XML files. It works with your favorite parser to provide idiomatic ways of navigating, searching, and modifying the parse tree. It commonly saves programmers hours or days of work.


**Parse using BeautifulSoap**




```
soap = BeautifulSoup(response.text, 'html.parser')
```

# **Print the HTML code that you get**

```
print(soap.prettify())
```

# <font color='Orange'>Extracting data</font> (0.7 POINTS)

From the printed HTML code search for the data you need for building the dataset, by the classes that are given above in the html tags listed.



**Try to select the first element from your attributes. Find out from the html code what part do you need.**

```
first_phone_brand = soap.find('h5', class_='mb-0')
if first_phone_brand:
    print("First Phone Brand Element:")
    print(first_phone_brand.text)


first_phone_type = soap.find('h3', class_='h5 font-weight-normal')
if first_phone_type:
    print("\nFirst Phone Type Element:")
    print(first_phone_type.text)


first_phone_price = soap.find('p', class_='h5 price')
if first_phone_price:
    print("\nFirst Phone Price Element:")
    print(first_phone_price.text)
```

**Add as many codeboxes as you need for selecting all the html elements**

```
phone_brands = [brand.text.strip() for brand in soap.find_all('h5', class_='mb-0')]
phone_types = [ptype.text.strip() for ptype in soap.find_all('h3', class_='h5 font-weight-normal')]
phone_prices = [price.text.strip() for price in soap.find_all('p', class_='h5 price')]

for brand, ptype, price in zip(phone_brands, phone_types, phone_prices):
    print("Brand:", brand)
    print("Type:", ptype)
    print("Price:", price)
    print()
```

**Create lists with the extracted data from the html elements that you selected**

```
phone_brands_list = []
phone_types_list = []
phone_prices_list = []

phone_brands_list.extend(phone_brands)
phone_types_list.extend(phone_types)
phone_prices_list.extend(phone_prices)

print(phone_prices_list)
```

**Print all lists that are containing the elements**

```
print("Phone Brands:")
print(phone_brands_list)
print("\nPhone Types:")
print(phone_types_list)
print("\nPhone Prices:")
print(phone_prices_list)
```

**Check if all lists are the same size**

```
if len(phone_brands) == len(phone_types) == len(phone_prices):
    print("All lists have the same size.")
else:
    print("Lists have different sizes.")
```

# <font color='Orange'>Create a Pandas DataFrame</font> (0.1 POINTS)


**Create the DataFrame from the lists you get and name your *Columns***

```
import pandas as pd

if len(phone_brands) == len(phone_types) == len(phone_prices):
    data = {
        'Brand': phone_brands,
        'Type': phone_types,
        'Price': phone_prices
    }

    df = pd.DataFrame(data)
```

**Print the dataset you obtain**

```
print(df)
```

#<font color='orange'>EDA</font> (0.2 POINTS)

## <font color='Orange'>Find statistic elements for the prices</font>
Like mean, median,  standard deviation, minimum, maximum

```
mean_price = df['Price'].mean();
median_price = df['Price'].median();
std_deviation = df['Price'].std();
min_price = df['Price'].min();
max_price = df['Price'].max();

print("Mean Price:", mean_price);
print("Median Price:", median_price);
print("Standard Deviation of Prices:", std_deviation);
print("Minimum Price:", min_price);
print("Maximum Price:", max_price);
```
